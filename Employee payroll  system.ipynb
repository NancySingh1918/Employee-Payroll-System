# PROJECT 04- EMPLOYEE PAYROLL SYSTEM
## Suggested Features
* Add Employees
* Calculate Salary
* Generate Payslips
* Search Employees
* Save Payroll Data

```python
import json

# We'll store employees in a list of dictionaries
employees = []

# -----------------------------
# Core Functions
# -----------------------------

def add_employee(emp_id, name, basic_salary, allowances=0, deductions=0):
    employee = {
        "emp_id": emp_id,
        "name": name,
        "basic_salary": basic_salary,
        "allowances": allowances,
        "deductions": deductions,
        "net_salary": 0,
        "payslip": ""
    }
    employees.append(employee)
    print(f" Employee {name} (ID: {emp_id}) added successfully!")

def calculate_salary(emp_id):
    for e in employees:
        if e["emp_id"] == emp_id:
            e["net_salary"] = e["basic_salary"] + e["allowances"] - e["deductions"]
            print(f" Net Salary for {e['name']} (ID: {emp_id}): {e['net_salary']}")
            return e["net_salary"]
    print(" Employee not found.")
    return None

def generate_payslip(emp_id):
    for e in employees:
        if e["emp_id"] == emp_id:
            payslip = f"""
            -------------------------------
            Payslip for {e['name']} (ID: {emp_id})
            -------------------------------
            Basic Salary : {e['basic_salary']}
            Allowances   : {e['allowances']}
            Deductions   : {e['deductions']}
            Net Salary   : {e['net_salary']}
            -------------------------------
            """
            e["payslip"] = payslip
            print(payslip)
            return payslip
    print(" Employee not found.")
    return None

def search_employee(emp_id):
    for e in employees:
        if e["emp_id"] == emp_id:
            print(f" Found: {e['name']} (ID: {emp_id}), Net Salary: {e['net_salary']}")
            return e
    print(" Employee not found.")
    return None

def save_payroll(filename="payroll.json"):
    with open(filename, "w") as f:
        json.dump(employees, f)
    print(" Payroll data saved successfully!")

def load_payroll(filename="payroll.json"):
    global employees
    try:
        with open(filename, "r") as f:
            employees = json.load(f)
        print(" Payroll data loaded successfully!")
    except FileNotFoundError:
        print(" No saved payroll data found.")

# -----------------------------
# Demo Usage
# -----------------------------
add_employee(1, "Anjali", 50000, allowances=5000, deductions=2000)
add_employee(2, "nancy", 60000, allowances=7000, deductions=3000)

calculate_salary(1)
calculate_salary(2)

generate_payslip(1)
generate_payslip(2)

search_employee(2)

save_payroll()
load_payroll()
```
